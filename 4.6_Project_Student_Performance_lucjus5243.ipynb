```python
# Step 3: Load the Student Performance Dataset

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# Load the dataset
df = pd.read_excel("StudentsPerformance.xlsx")

# Display the first five rows
print("First Five Rows:")
display(df.head())

# Display dataset information
print("\nDataset Information:")
df.info()

# Check for missing values
print("\nMissing Values:")
print(df.isnull().sum())

# Display summary statistics
print("\nSummary Statistics:")
display(df.describe())
```

```python
# Step 4: One-Hot Encoding

target = "math score"

X = df.drop(columns=[target])
y = df[target]

X = pd.get_dummies(X, drop_first=True, dtype=int)

print("Encoded Dataset:")
display(X.head())

print("\nTotal Features:", X.shape[1])

print("\nFeature Names:")
print(X.columns.tolist())
```

```python
# Step 5: Train and Test Linear Regression

from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression

# Split data: 80% training, 20% testing
X_train, X_test, y_train, y_test = train_test_split(
    X, y,
    test_size=0.20,
    random_state=42
)

# Train the Linear Regression model
model = LinearRegression()
model.fit(X_train, y_train)

print("Training Samples:", len(X_train))
print("Testing Samples:", len(X_test))
print("Linear Regression model trained successfully!")
```

```python
# Step 6: Model Evaluation

from sklearn.metrics import (
    mean_absolute_error,
    mean_squared_error,
    r2_score
)

# Predict math scores
y_pred = model.predict(X_test)

# Calculate evaluation metrics
mae = mean_absolute_error(y_test, y_pred)
mse = mean_squared_error(y_test, y_pred)
r2 = r2_score(y_test, y_pred)

# Display results
print("Model Evaluation Results")
print("------------------------")
print(f"Mean Absolute Error (MAE): {mae:.4f}")
print(f"Mean Squared Error (MSE): {mse:.4f}")
print(f"R-Squared Value: {r2:.4f}")
```

```python
# Step 7: Feature Importance

import pandas as pd
import matplotlib.pyplot as plt

# Calculate feature importance
importance = pd.DataFrame({
    "Feature": X.columns,
    "Coefficient": model.coef_
})

# Calculate absolute importance
importance["Absolute_Importance"] = (
    importance["Coefficient"].abs()
)

# Rank features by importance
importance = importance.sort_values(
    by="Absolute_Importance",
    ascending=False
)

# Display the ranking
print("Feature Importance Ranking:")
display(importance)

# Create a graph
plt.figure(figsize=(11, 7))

plt.barh(
    importance["Feature"],
    importance["Coefficient"]
)

plt.xlabel("Regression Coefficient")
plt.ylabel("Feature")
plt.title("Feature Importance for Predicting Math Scores")
plt.axvline(0, color="black", linewidth=0.8)
plt.tight_layout()
plt.show()
```
